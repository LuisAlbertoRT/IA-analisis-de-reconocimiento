```python
# =========================================================
# Captura de Kinect v1 (Xbox 360) usando PyKinect2 / PyKinectRuntime
# Kernel: Python 3.9 (kinect_env)
# =========================================================
from pykinect2 import PyKinectRuntime, PyKinectV2
import cv2
import numpy as np
from IPython.display import display, Image, clear_output
import time

# Inicializar sensores (Color + Profundidad)
kinect = PyKinectRuntime.PyKinectRuntime(
    PyKinectV2.FrameSourceTypes_Color | PyKinectV2.FrameSourceTypes_Depth
)

print("Iniciando flujo de Kinect v1... Presiona 'Interrupt Kernel' para detener.")

try:
    while True:
        # 1. Obtener frame a color (RGB)
        if kinect.has_new_color_frame():
            color_frame = kinect.get_last_color_frame()
            color_img = color_frame.reshape((1080, 1920, 4))[:, :, :3] # Convertir RGBA a RGB
            color_img = cv2.resize(color_img, (640, 360))
        else:
            color_img = None

        # 2. Obtener frame de profundidad
        if kinect.has_new_depth_frame():
            depth_frame = kinect.get_last_depth_frame()
            depth_img = depth_frame.reshape((424, 512)).astype(np.uint8)
            depth_colormap = cv2.applyColorMap(depth_img, cv2.COLORMAP_JET)
            depth_colormap = cv2.resize(depth_colormap, (640, 360))
        else:
            depth_colormap = None

        # 3. Mostrar combinados si ambos frames están listos
        if color_img is not None and depth_colormap is not None:
            combined = np.hstack((color_img, depth_colormap))
            _, buffer = cv2.imencode('.jpg', combined)
            
            clear_output(wait=True)
            display(Image(data=buffer))
            
        time.sleep(0.03)

except KeyboardInterrupt:
    print("\nTransmisión finalizada.")
    kinect.close()
```
